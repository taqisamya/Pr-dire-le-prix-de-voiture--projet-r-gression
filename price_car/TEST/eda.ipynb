```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"data\train_post_modif_median.csv")
df
```

```python
plt.figure(figsize=(10,5))                                          # Create new figure
df["Year"].hist(bins=40)                      # Histogram of target
plt.title(" prix en fonction de l'année)") # Plot title
plt.xlabel("Year")                                  # X label
plt.ylabel("Price")                                   # Y label

plt.xticks(range(int(df["Year"].min()), int(df["Year"].max())+1), rotation=90) #forcer tout les années à apparaitre
plt.show() 
```

```python
plt.figure(figsize=(10,5))                                          # Create new figure
df["Brand"].hist(bins=40)                      # Histogram of target
plt.title("Évolution de prix en fonction de la marque)") # Plot title
plt.xlabel("Brand")                                  # X label
plt.ylabel("Price") 
plt.xticks(rotation=90)   
plt.tight_layout()                               # Y label
plt.show() 
```

```python
plt.figure(figsize=(12,6))

df["Brand"].value_counts().plot(kind="bar", colormap="pink")

plt.title("Nombre de voitures par marque")
plt.xlabel("Brand")
plt.ylabel("Count")

plt.xticks(rotation=90)   # <-- Rotation verticale
plt.tight_layout()        # Ajuste automatiquement l'espace

plt.show()
```

```python
df["Fuel_Type"]=pd.factorize(df["Fuel_Type"])[0]
df["Transmission"]=pd.factorize(df["Transmission"])[0]
df["Owner_Type"]=pd.factorize(df["Owner_Type"])[0]
df["Brand"]=pd.factorize(df["Brand"])[0]
df["Model"]=pd.factorize(df["Model"])[0]
df
```

```python
correlation= df.corr()
sns.heatmap(correlation, annot=True , cmap="coolwarm")
plt.title("la matrice de corrélation ")
plt.legend()
plt.show()
```
