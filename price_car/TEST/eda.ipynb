```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"..\data\train_clean.csv")
df
```

```python
plt.figure(figsize=(10,5))                                          # Create new figure
df["Year"].hist(bins=40)                      # Histogram of target
plt.title(" prix en fonction de l'année") # Plot title
plt.xlabel("Year")                                  # X label
plt.ylabel("Price")                                   # Y label

plt.xticks(range(int(df["Year"].min()), int(df["Year"].max())+1), rotation=90) #forcer tout les années à apparaitre
plt.show() 
```

```python
plt.figure(figsize=(10,5))                                          # Create new figure
df["Brand"].hist(bins=40)                      # Histogram of target
plt.title("Évolution de prix en fonction de la marque") # Plot title
plt.xlabel("Brand")                                  # X label
plt.ylabel("Price") 
plt.xticks(rotation=90)   
plt.tight_layout()                               # Y label
plt.show() 
```

```python
plt.figure(figsize=(12,6))

df["Brand"].value_counts().plot(kind="bar", colormap="pink")

plt.title("Nombre de voitures par marque")
plt.xlabel("Brand")
plt.ylabel("Count")

plt.xticks(rotation=90)   # <-- Rotation verticale
plt.tight_layout()        # Ajuste automatiquement l'espace

plt.show()
```

```python
df["Fuel_Type"]=pd.factorize(df["Fuel_Type"])[0]
df["Transmission"]=pd.factorize(df["Transmission"])[0]
df["Owner_Type"]=pd.factorize(df["Owner_Type"])[0]
df["Brand"]=pd.factorize(df["Brand"])[0]
df["Model"]=pd.factorize(df["Model"])[0]
df["Location"]=pd.factorize(df["Location"])[0]
df
```

```python
correlation= df.corr()
plt.figure(figsize=(10,8))
sns.heatmap(correlation, annot=True , cmap="coolwarm")
plt.title("la matrice de corrélation ")
plt.legend()
plt.show()
```

```python
sns.pairplot(df, 
             vars=['Fuel_Type','Year','Mileage'],
             hue='Price')
```

```python
sns.catplot(x='Transmission', y='Price',data=df, hue='Power', kind='swarm')
```

```python
sns.histplot(data=df, x='Price',hue='Brand', multiple ='stack')
```

```python
plt.figure(figsize=(10,8))
sns.lineplot(data=df, x='Year', y='Price',hue='Brand', style='Fuel_Type', estimator='mean')
plt.title=("Evolution moyenne du prix par la marque , Fuel_Type et l'annee")
plt.xlabel("Year")
plt.ylabel("Mean Price")
plt.legend(bbox_to_anchor=(1.05,1))
plt.tight_layout()
plt.show()
```

```python
g = sns.FacetGrid(df, col="Fuel_Type", hue="Brand", col_wrap=2)
g.map(sns.lineplot, "Year", "Price", estimator="mean")
g.add_legend()
plt.show()
```

```python
import plotly.express as px
df_grouped = df.groupby(
        ["Year", "Brand", "Fuel_Type"]
    )["Price"].mean().reset_index()

fig = px.line(
        df_grouped,
        x="Year",
        y="Price",
        color="Brand",
        line_dash="Fuel_Type",
        title="Évolution moyenne du prix par Marque et Fuel"
    )

fig.update_layout(
        template="plotly_white",
        height=600
    )
```
