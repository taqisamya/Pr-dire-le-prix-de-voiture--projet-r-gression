```python
def iqr(data, col_name):
    q1 = data[col_name].quantile(0.25)
    q3 = data[col_name].quantile(0.75)
    iqr_ = q3 - q1

    upper_lim = q3 + 1.5*iqr_
    lower_lim = q1 - 1.5*iqr_

    outliers_above = data[data[col_name]>upper_lim]
    outliers_below = data[data[col_name]<lower_lim]

    print("\n", col_name, ":")

    print("Outliers above :",outliers_above[col_name].count())
    print("Outliers below :",outliers_below[col_name].count())


for col in data2.select_dtypes(include=[np.number]).columns:
    iqr(data2, col)
```

```python
def replace_outliers_by_median(data, col_name):  
    q1 = data[col_name].quantile(0.25)
    q3 = data[col_name].quantile(0.75)
    iqr_ = q3 - q1

    upper_lim = q3 + 1.5*iqr_
    lower_lim = q1 - 1.5*iqr_

    median = data[col_name].median()

    data.loc[data[col_name]>upper_lim, col_name] = median
    data.loc[data[col_name]<lower_lim, col_name] = median

    return data

data_transform = data2.copy()

for col in data_transform.select_dtypes(include=[np.number]).columns:
    data_transform = replace_outliers_by_median(data_transform, col)

```

```python
from sklearn.pipeline import Pipeline                  # Pipeline to prevent leakage
from sklearn.compose import ColumnTransformer          # Different preprocessing per column type
from sklearn.preprocessing import StandardScaler, RobustScaler       # Feature scaling (important for SVR)
from sklearn.preprocessing import OneHotEncoder        # Convert categorical to one-hot vectors
from sklearn.impute import SimpleImputer
```

```python
numeric_features = data2.select_dtypes(include=[np.number]).columns.tolist()
categorical_features = data2.select_dtypes(exclude=[np.number]).columns.tolist()

numeric_transformer = Pipeline(steps=[
                                                     # Fill missing numeric values with median (computed on TRAIN only)
    ("scaler", RobustScaler())                       # Standardize features (fit on TRAIN only) RobustScaler pour gérer le outliers
])
categorical_transformer = Pipeline(steps=[
                                                             # Fill missing categories with most frequent (TRAIN only)
    ("onehot", OneHotEncoder(handle_unknown="ignore"))     # One-hot encode categories (fit on TRAIN only)
])

preprocess = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),     # Apply numeric pipeline to numeric columns
        ("cat", categorical_transformer, categorical_features)  # Apply categorical pipeline to categorical columns
    ]
)
```
