```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
data = pd.read_csv('..\\data\\train.csv')
data.head()
```

```python
data["Model"] = data["Name"].str.split()
data["Brand"] = data["Model"].str[0]
data.head()
```

```python
data["Model"] = data["Model"].str[1:]
data.head()
```

```python
data["Model"] = data["Model"].str.join(' ')
data.head()
```

```python
data = data.drop(["Name", "Location", "Engine", "Seats", "New_Price"], axis=1)
data.head()
```

```python
data["Mileage"] = data["Mileage"].str.split()
data["Mileage"] = data["Mileage"].str[0]
data["Power"] = data["Power"].str.split()
data["Power"] = data["Power"].str[0]
data.head()
```

```python
data["Price"] = data["Price"]*929.9
```

```python
data.head(20)
```

```python
data["Brand"] = data["Brand"].str.upper()
data.head()
```

```python
data["Model"] = data["Model"].str.upper()
data.head()
```

```python
data["Mileage"] = pd.to_numeric(data["Mileage"], errors='coerce')
data["Power"] = pd.to_numeric(data["Power"], errors='coerce')
data.info()
```

```python
data["Power"] = data["Power"].fillna(data["Power"].median(), inplace=True)
data["Mileage"] = data["Mileage"].fillna(data["Mileage"].median(), inplace=True)
data.info()
```

```python
# Fonction pour visualiser les outliers : 
def iqr(data, col_name):
    q1 = data[col_name].quantile(0.25)
    q3 = data[col_name].quantile(0.75)
    iqr_ = q3 - q1

    upper_lim = q3 + 1.5*iqr_
    lower_lim = q1 - 1.5*iqr_

    outliers_above = data[data[col_name]>upper_lim]
    outliers_below = data[data[col_name]<lower_lim]

    print("\n", col_name, ":")

    print("Outliers above :",outliers_above[col_name].count())
    print("Outliers below :",outliers_below[col_name].count())
```

```python
for col in data[["Kilometers_Driven", "Mileage", "Power"]].columns:
    iqr(data, col)
```

```python
# Fonction pour remplacer les outliers par la médiane : 
def replace_outliers_by_median(data, col_name):  
    q1 = data[col_name].quantile(0.25)
    q3 = data[col_name].quantile(0.75)
    iqr_ = q3 - q1

    upper_lim = q3 + 1.5*iqr_
    lower_lim = q1 - 1.5*iqr_

    median = data[col_name].median()

    data.loc[data[col_name]>upper_lim, col_name] = median
    data.loc[data[col_name]<lower_lim, col_name] = median

    return data
```

```python
for col in data[["Kilometers_Driven", "Mileage", "Power"]].columns:
    replace_outliers_by_median(data, col)
    iqr(data, col)
```
