```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
data = pd.read_csv("..\\data\\train_post_modif_median.csv")
data.head()
```

```python
corr = data.drop(["Fuel_Type", "Transmission", "Owner_Type", "Model","Brand"], axis=1).corr()
f, ax = plt.subplots(figsize=(12, 8))
mat = sns.heatmap(corr, cmap='coolwarm', annot=True, linewidths=1.2)
mat.axes.set_title("Matrice de corrélation entre les features", fontsize=15)
plt.show()
```

```python
data["Brand"] = pd.factorize(data["Brand"])[0]
data["Fuel_Type"] = pd.factorize(data["Fuel_Type"])[0]
data["Transmission"] = pd.factorize(data["Transmission"])[0]
data["Owner_Type"] = pd.factorize(data["Owner_Type"])[0]
data["Model"] = pd.factorize(data["Model"])[0]
```

```python
corr = data.corr()
f, ax = plt.subplots(figsize=(12, 9))
mat = sns.heatmap(corr, cmap='coolwarm', annot=True, linewidths=1.2)
mat.axes.set_title("Matrice de corrélation entre les features", fontsize=15)
plt.show()
```
