```python
import numpy as np                           
import pandas as pd                                 
import matplotlib.pyplot as plt 
```

```python
data=pd.read_csv("..\\data\\train_clean.csv")
data.head()
```

```python
df_moyen = data.groupby('Year')['Price'].mean()

plt.figure(figsize=(12,8))
plt.plot(df_moyen.index, df_moyen.values, color='navy')
plt.title('Prix moyen par année', fontsize=14)
plt.xlabel('Année', fontsize=12)
plt.ylabel('Prix moyen (en centaine de milliers de rupees)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.xticks(df_moyen.index)

plt.show()
```

```python
df_moyen = data.groupby('Fuel_Type')['Price'].mean().sort_values()

plt.figure(figsize=(12,8))
plt.bar(df_moyen.index, df_moyen.values, color='orange', edgecolor='black')
plt.title('Prix moyen par type de carburant', fontsize=14)
plt.xlabel('Carburant', fontsize=12)
plt.ylabel('Prix moyen (en centaine de milliers de rupees)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.xticks(df_moyen.index)

plt.show()
```

```python
df_moyen = data.groupby('Brand')['Price'].mean().sort_values()

plt.figure(figsize=(28,8))
plt.bar(df_moyen.index, df_moyen.values, color='navy', edgecolor='navy')
plt.title('Prix moyen par marque', fontsize=14)
plt.xlabel('Marque', fontsize=12)
plt.ylabel('Prix moyen (en centaine de milliers de rupees)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.xticks(df_moyen.index)

plt.show()
```

```python
def evo_prix_par_marque(brand:str):
    df_moyen_marque = data[data["Brand"]==brand].groupby('Year')['Price'].mean()

    plt.figure(figsize=(12,8))
    plt.plot(df_moyen_marque.index, df_moyen_marque.values, color='navy')
    plt.title(f'Prix moyen des véhicules {brand} par année', fontsize=14)
    plt.xlabel('Année', fontsize=12)
    plt.ylabel('Prix moyen (en centaine de milliers de rupees)', fontsize=12)
    plt.grid(axis='y', linestyle='--', alpha=0.7)

    plt.xticks(df_moyen_marque.index)

    plt.show()

evo_prix_par_marque('MARUTI')
```

```python
evo_prix_par_marque('BMW')
```

```python
evo_prix_par_marque('AUDI')
```

```python
def prix_marque_par_carburant(brand:str):
    df_moyen = data[data["Brand"]==brand].groupby('Fuel_Type')['Price'].mean().sort_values()

    plt.figure(figsize=(12,8))
    plt.bar(df_moyen.index, df_moyen.values, color='orange', edgecolor='black')
    plt.title(f'Prix moyen des véhicules {brand} par type de carburant', fontsize=14)
    plt.xlabel('Carburant', fontsize=12)
    plt.ylabel('Prix moyen (en centaine de milliers de rupees)', fontsize=12)
    plt.grid(axis='y', linestyle='--', alpha=0.7)

    plt.xticks(df_moyen.index)

    plt.show()

prix_marque_par_carburant('MARUTI')
```
